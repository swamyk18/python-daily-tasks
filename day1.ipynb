Problem-1: Find Two Numbers with a Given Sum

Input:  nums = [2, 7, 11, 15], target = 9
Output: [0, 1]

Input:  nums = [3, 2, 4], target = 6
Output: [1, 2]

Input:  nums = [3, 3], target = 6
Output: [0, 1]

Input:  nums = [1, 5, 8, 10], target = 20
Output: None

Input:  nums = [], target = 5
Output: None

```python
def two_sum(nums,target):

    if len(nums)<2:
        return None
    i=0
    j=len(nums)-1
    while i<j:
        sum=nums[i]+nums[j]
        if sum==target:
            return {i,j}
        elif sum<target:
            i+=1
        else:
            j-=1
    
            
    return None


nums=[1, 5, 8, 10]
target=20
print(f'{two_sum(nums,target)}')
```

Problem-3: Find the First Non-Repeating Character

You are given a string. Write a Python function to find the first character that appears only once in the string.

If every character is repeated, return None.

Examples
Example 1:
Input:  "programming"
Output: 'p'

```python
def nonRepeatingChar(str):
    freq={}
    for ch in str:
        if ch not in freq:
            freq[ch]=1
        else:
            freq[ch]+=1

    for ch,val in freq.items():
        if val==1:
            return ch
    return None

str="programming"
str="aabbcdd"
str="aabbcc"
print(nonRepeatingChar(str))
```
